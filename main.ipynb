Presently, the calculations are done based on regional crime indexes hardcore coded into the system.

```python
regions = dict(        # Making use of a dictonary (Key=>value pairs). Could also use {}. data = {"Key" : "Value", "Key" : "Value"}
    far_north=67, 
    north=46, 
    adamawa=60, 
    west=69, 
    north_west=83, 
    south_west=71, 
    littoral=75, 
    center=77, 
    south=33, 
    east=47
)
```

```python
# user input.
while True:
    location = input("Enter the region you wish to go to (It must be one of the 10 regions in Cameroon: ")
    location=location.lower().replace(" ","_")      #clean text
    if location in regions:
        #print(location)
        break;
    print("\nIt seems the region you entered isn't in Cameroon!\n")
```

```python
# Color codes
GREEN = "\033[32m"
YELLOW = "\033[33m"
RED = "\033[31m"
RESET = "\033[0m"   # Terminal default color

# Risk Calculation Logic
if regions[location] >= 70:
    risk = "high"
    COLOR = RED
elif regions[location] >= 40:
    risk = "moderate"
    COLOR = YELLOW
else:
    risk = "low"
    COLOR = GREEN
```

```python
# Calculated Output
print(f"\nYour Destination: {location.replace("_", " ").title()} \nRisk Level: {COLOR}{risk.title()}{RESET}")
```
